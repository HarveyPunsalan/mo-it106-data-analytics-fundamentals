```python
import pandas as pd

# load the cleaned product list
df = pd.read_csv('MotorPH_Product_List_Cleaned.csv')

# show the first few rows
df.head()
```

```python
# 3.a total number of products
print('Total number of products:', len(df))
```

```python
# 3.b counts by product type
print(df['Product Category'].value_counts())
```

```python
# 3.c unit price statistics
print('Average (mean) price:', format(df['Unit Price'].mean(), ',.2f'))
print('Cheapest price:', format(df['Unit Price'].min(), ','), '-', df.loc[df['Unit Price'].idxmin(), 'Product Name'])
print('Most expensive price:', format(df['Unit Price'].max(), ','), '-', df.loc[df['Unit Price'].idxmax(), 'Product Name'])

# also show the spread of the prices
print(df['Unit Price'].describe())
```

```python
# 3.d total inventory cost
print('Total inventory cost:', format(df['Unit Price'].sum(), ','))
```

## Observations

Scooters make up most of the list with 12 of the 50 products (24%). Naked bikes come next with 7, and the rest of the types only have 1 or 2 products each.

The prices go from ₱48,000 to ₱995,000, so the most expensive unit costs around 20 times more than the cheapest one. The list has both entry level and premium motorcycles.

The average price (₱257,872) is higher than the median (₱204,950). A few expensive models pull the average up, so a typical product is cheaper than the average.

Half of the products cost less than ₱204,950, and the middle half of the prices is between ₱126,425 and ₱330,250. Most of the list is made up of budget and mid-range motorcycles.

The total inventory cost is ₱12,893,600. The expensive models are few but each one costs a lot, so the scooter and underbone range can be stocked in bigger quantities while the expensive units are better kept in smaller numbers.
